```python
from markov_volatility import MarkovVolatility
import warnings
warnings.filterwarnings('ignore')
```

```python
# Initialize the model with your data
model = MarkovVolatility('data/micro/candleData/bitcoin_candles.csv', train_test_split=0.75, init_date='2014-01-01', embargo_pct=0.01)

# Fit the Markov model
model.fit_markov()
```

```python

# Get volatility state for a specific date
state = model.get_volatility_state("2015-12-30")  # Returns 'high' or 'low'

state
```

```python
# Get all volatility states
all_states = model.get_volatility_state()  # Returns Series with dates and states
all_states
```

```python

# Visualize the volatility regimes
model.plot_volatility()
```

```python
# Save the model for later use
model.save_model()
```
