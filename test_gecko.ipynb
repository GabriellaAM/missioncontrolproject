```python
from pycoingecko import CoinGeckoAPI
from dotenv import load_dotenv
import os
from geckoAPI.utils import datetime_to_unix, parse_gecko_ohlcv, parse_gecko_prices

load_dotenv()

# Exemplo de lista de ativos (IDs no CoinGecko)
coin_ids = ["bitcoin", "ethereum", "cardano"]  
# Ajuste aqui para os ativos que você quer buscar.

# Inicializa a API
cg = CoinGeckoAPI(api_key=os.getenv('GECKO_API_KEY'))
```

```python
data = cg.get_coin_ohlc_by_id_range(
            id='heyanon', 
            vs_currency='usd',
            from_timestamp=datetime_to_unix('2025-01-01'),
            to_timestamp=datetime_to_unix('2025-01-07'),
            interval='daily'
        )
parse_gecko_ohlcv(data)
```

```python
data = cg.get_coin_market_chart_by_id(
            id='bitcoin', 
            vs_currency='usd',
            days=100000,
            interval='daily'
        )

parse_gecko_prices(data)
```
